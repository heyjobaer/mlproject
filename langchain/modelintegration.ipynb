```python
from langchain.chat_models import init_chat_model

model = init_chat_model("openai/gpt-oss-120b", model_provider="groq")
response = model.invoke("Write 200 word paragrah about what is artificial intellgence?")
print(response.content)
```

```python
model.stream("Write 200 words paragrah on artificial intellgence?")
```
